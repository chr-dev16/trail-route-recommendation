```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("../../data/processed/dem_features.xlsx")

row = df.iloc[0]

thresholds = [0, 1, 2, 3, 5, 7.5, 10, 12.5, 15]

ascent = [
    row[f"total_ascent_{t}"]
    for t in thresholds
]

descent = [
    row[f"total_descent_{t}"]
    for t in thresholds
]

plt.figure(figsize=(10, 5))

plt.plot(thresholds, ascent, marker="o", label="Ascent")
plt.plot(thresholds, descent, marker="o", label="Descent")

plt.xlabel("Elevation threshold (m)")
plt.ylabel("Elevation gain/loss (m)")
plt.title(f"Threshold sensitivity - {row['route']}")
plt.grid(True)
plt.legend()

plt.show()
```
